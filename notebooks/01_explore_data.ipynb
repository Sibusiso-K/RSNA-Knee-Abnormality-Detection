# RSNA Knee — data exploration

First pass: what files came down, how big is the label table, how imbalanced are the 12 targets, and what does a single MRI study look like.

```python
from pathlib import Path

import pandas as pd

DATA = Path('..') / 'data'
sorted(p.name for p in DATA.iterdir())
```

```python
train = pd.read_csv(DATA / 'train.csv')
print(train.shape)
train.head()
```

```python
TARGETS = [
    'ACL', 'MCL', 'Medial Meniscus', 'Lateral Meniscus', 'Medial OA',
    'Lateral OA', 'PF OA', 'Effusion', 'Synovitis', "Baker's", 'Contusion',
    'Fracture',
]
train[TARGETS].mean().sort_values(ascending=False)
```

```python
# Peek at one study's DICOM series
import pydicom

dicoms = sorted((DATA / 'train_images').rglob('*.dcm'))[:5]
for d in dicoms:
    ds = pydicom.dcmread(d, stop_before_pixels=True)
    print(d.relative_to(DATA), ds.get('SeriesDescription'), ds.get('Rows'), ds.get('Columns'))
```
